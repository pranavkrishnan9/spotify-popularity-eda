```python
#imports
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
#read dataset to var
spotify = pd.read_csv("../content/sample_data/spotify_dataset.csv")
spotify.head()
```

```python
#input var
# Reminder: encapsulate input features even if using single predictor since sklearn model is multiple linear regression by default
X = spotify[['loudness']]
Y = spotify.popularity
```

```python
#Comparing gradient descent w/ linear regression model
reg = LinearRegression().fit(X,Y)
reg.coef_
```

```python
reg.intercept_
```

```python
#Gradient Descent implementation -- should match reg coef and intercept
X = spotify[['loudness']]
Y = spotify.popularity
ones = np.ones((len(spotify), 1))
attributes = X.values
x1 = np.column_stack((ones, attributes))

x1_transpose = np.transpose(x1)
y = np.array(Y)
B0 = 0
B1 = 0
B = np.array((B0, B1))
learning = []

for i in range(100000):
  #popularity prediction
  y1 = x1 @ B

  #cost calculation
  cost = 1/len(y1) * np.sum((y1 - y) ** 2)


  #gradient calculation with cost iteration
  gradient = 2/len(y1) * (x1_transpose @ (y1-y))

  #learning rate calibration
  alpha = 0.001
  B = B - alpha*gradient
  learning.append(cost)

learning
B
np.var(y)
```
